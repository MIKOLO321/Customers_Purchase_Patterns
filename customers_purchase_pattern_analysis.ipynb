```python
# To load the dataset using pandas

df = pd.read_csv("C:/Users/USER MICHAEL$$/Downloads/customer_purchase_pattern.csv")
```

```python
# checking null values

df.isnull().sum()
```

```python
# To find the median and replace it with the missing value within each category
df['Rating score'] =df.groupby('Category')['Rating score'].transform(lambda x: x.fillna(x.median()))
```

```python
# checking the null values again

df.isnull().sum()
```

```python
# changing all the columns to snake_case 

df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')
df=df.rename(columns={'purchase_amount_(usd)':'amount_spent'})
```

```python
# check the columns

df.columns
```

```python
# create age_bracket

labels = ['teenager', 'adult', 'mature_adult', 'senior']
df['age_bracket']= pd.qcut(df['age'], q= 4, labels = labels)
```

```python
df[['age', 'age_bracket']].head()
```

```python
# create another column frequency_purchase_days

mapping_frequency = {
    'Weekly': 7,
    'Bi_Weekly': 14,
    'Every 3 Months': 90,
    'Monthly': 30,
    'Quarterly': 90,
    'Fortnightly': 14,
    'Annually': 365
}
df['frequency_purchase_days'] = df['purchase_frequency'].map(mapping_frequency)
```

```python
df[['purchase_frequency','frequency_purchase_days']].head(10)
```

```python
df[['discount_used', 'promo_code_applied']].head()
```

```python
(df['discount_used'] == df['promo_code_applied']).all()
```

```python
# since both discount_used and promo_code_applied have same data, lets drop one of them

df = df.drop('promo_code_applied', axis =1)
```
